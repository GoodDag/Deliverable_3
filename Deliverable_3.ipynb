Character class

```python
d8 = Dice(sides=8)
sword = Weapon(name='sword', damage_dice=d8)

aria = Character(name="Aria", hp_max=100, damage=10)
kura = Character(name="Kura", hp_max=100, damage=20)
swordsman = Character(name="Swordsman", hp_max=100, weapon=sword)

print('introductions:')
aria.introduce()
print(aria.__repr__())
print(aria)
print(swordsman.__repr__())
print(swordsman)
print()

print('Aria starting hp:', aria.hp)
aria.take_damage(kura)
print('hp after taking damage:', aria.hp, 'out of', aria.hp_max)
aria.heal()
print('hp after 10 heal:', aria.hp)
kura.attack(aria)
print('hp after being attacked by kura:', aria.hp)
swordsman.attack(aria)
print('hp after being attacked by swordsman:', aria.hp)
kura.attack(aria)
kura.attack(aria)
kura.attack(aria)
print('Is aria alive?', aria.is_alive, 'hp:', aria.hp)
kura.attack(aria)
print('Is aria alive?', aria.is_alive, 'hp:', aria.hp)
```

Battle class

Default character vs. default mage test round:

```python
default_char = Character(name='John Doe')
default_mage = Mage(name='Magic Jane Doe')

print('Test round:')
print(default_char)
print(default_mage)
print()
test_round = Battle(hero=default_char, enemy=default_mage)
test_round.play_round()
print(default_char)
print(default_mage)
print()
```

Default warrior vs. spearman:

```python
default_warrior = Warrior(name='Sturdy John Doe')
d20 = Dice(sides=20)
spear = Weapon(name='Spear', damage_dice=d20)
spearman = Character(name='Spearman', weapon=spear)

print('Battle 1:')
print(default_warrior)
print(spearman)
print()
battle1 = Battle(hero=default_warrior, enemy=spearman)
print(battle1.simulate_battle())
```

High damage vs. high armor:

```python
high_damage = Mage(name='High damage', hp_max=100, damage=15)
high_armor = Warrior(name='High armor', hp_max=100, damage=5, armor=15)

print('Battle 2:')
print(high_damage)
print(high_armor)
print()
battle2 = Battle(hero=high_damage, enemy=high_armor)
print(battle2.simulate_battle())
```

Sorcerer vs. archer

```python
d10 = Dice(sides=10)
staff = Weapon(name='staff', damage_dice=d10)
sorcerer = Mage(name='Sorcerer', damage=5, weapon=staff)

bow = Weapon(name='Bow', damage_dice=d10)
archer = Character(name='Archer', weapon=bow)

print('Battle 3:')
print(sorcerer)
print(archer)
print()
battle3 = Battle(hero=sorcerer, enemy=archer)
print(battle3.simulate_battle())
```
