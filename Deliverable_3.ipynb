Character class

```python
d8 = Dice(sides=8)
sword = Weapon(name='sword', damage_dice=d8)

aria = Character(name="Aria", hp_max=100, damage=10)
kura = Character(name="Kura", hp_max=100, damage=20)
swordsman = Character(name="Swordsman", hp_max=100, weapon=sword)

print('introductions:')
aria.introduce()
print(aria.__repr__())
print(aria)
print(swordsman.__repr__())
print(swordsman)
print()

print('Aria starting hp:', aria.hp)
aria.take_damage(kura)
print('hp after taking damage:', aria.hp, 'out of', aria.hp_max)
aria.heal()
print('hp after 10 heal:', aria.hp)
kura.attack(aria)
print('hp after being attacked by kura:', aria.hp)
swordsman.attack(aria)
print('hp after being attacked by swordsman:', aria.hp)
kura.attack(aria)
kura.attack(aria)
kura.attack(aria)
print('Is aria alive?', aria.is_alive, 'hp:', aria.hp)
kura.attack(aria)
print('Is aria alive?', aria.is_alive, 'hp:', aria.hp)
```
